```python
# missing values
print("old data frame length:", len(df))
df.isna().sum()

df = df.dropna(axis=0)
print("New data frame length:", len(df))
```

```python
# Handle duplicate rows
# print(df.duplicated())
len(df.loc[df.duplicated()])
```

```python
def zscore_outliers(feature):
    upper_limit = (df[feature].mean()) + (3*df[feature].std())
    lower_limit = df[feature].mean()-3*df[feature].std()
    print("upper limit:",upper_limit)
    print("lower limit:",lower_limit)
    print(df.loc[(df[feature]>upper_limit) | (df[feature]<lower_limit)])
      
```

```python
zscore_outliers('age')
```

```python
zscore_outliers('employ')
```
