```python
df['Income'] = df['Income'].fillna(df['Income'].median()) 
```

```python
#converting categorical to numeric

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Creating an instance of OneHotEncoder
onehotencoder = OneHotEncoder()

# Fitting and transforming the 'Marital_Status' column
marital_status_encoded = onehotencoder.fit_transform(df[['Marital_Status']]).toarray()
marital_status_encoded

# # Creating a DataFrame with the encoded values
marital_status_df = pd.DataFrame(marital_status_encoded, columns=onehotencoder.get_feature_names_out(['Marital_Status']))
marital_status_df

# # Concatenating the new DataFrame with the original DataFrame
df = pd.concat([df, marital_status_df], axis=1)

# # Dropping the original 'Marital_Status' column
df.drop('Marital_Status', axis=1, inplace=True)

# df.head()
```

```python
# Fitting and transforming the 'education' column
Education_encoded = onehotencoder.fit_transform(df[['Education']]).toarray()
Education_encoded

# # Creating a DataFrame with the encoded values
Education_encoded_df = pd.DataFrame(Education_encoded, columns=onehotencoder.get_feature_names_out(['Education']))
Education_encoded_df

# # # Concatenating the new DataFrame with the original DataFrame
df = pd.concat([df, marital_status_df], axis=1)

# # # Dropping the original 'Marital_Status' column
df.drop('Education', axis=1, inplace=True)
# df.head()
```
